# Anatomia de una grafica de Matplotlib

## Figura

Es el **lienzo** sobre el cual dibujaremos nuestra grafica. En realidad es un objeto tipo `Figure`:

![](../img/eda/matplotlib_figure.png)

A medida que elaboremos la grafica iremos añadiendo **hijos** a este lienzo (ejes, lineas, puntos, grillas, leyendas, etc.)

## Ejes

Es el objeto (tipo `axes`) donde realmente dibujaremos las graficas

![](../img/eda/matplotlib_1_axes.png)

o con multiples ejes:

![](../img/eda/matplotlib_4_axes.png)

Y cada par de ejes de la fugura tienen unos **atributos**
Algunos ejemplos de atributos son:

- Los **axis** o ejes individuales (ejes horizontales, ejes verticales)
- Las etiquetas de los ejes (que podemos fijar con los metodos `set_xlabel()` y `set_ylabel()`)

## Ejes individuales

Son objetos tipo `axis` que a su vez son hijos tanto de los ejes como de la figura
