# Generando mulples graficos

### 1. Generando arreglos de Axes

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# generar una grilla de 2 filas y 1 columna
fig, axs = plt.subplots(2, 1)
```

```python
axs
```

```python
# generar una grilla de 2 filas y 3 columna
fig, axs = plt.subplots(2, 3)
```

```python
axs
```

```python
# crear un arreglo x (variables independientes)
x = np.linspace(-5, 5, 100)
# crear grilla
fig, axs = plt.subplots(2, 3)
# iterar sobre la grilla para generar graficos
for i in range(2):
    for j in range(3):
        axs[i, j].plot(x, x ** (i + j))
        axs[i, j].set_title(f"i, j = {i}, {j}")
```

```python
# crear un arreglo x (variables independientes)
x = np.linspace(-5, 5, 100)
# crear grilla
fig, axs = plt.subplots(2, 3)
fig.subplots_adjust(hspace=0.5, wspace=0.5) 
# iterar sobre la grilla para generar graficos
for i in range(2):
    for j in range(3):
        axs[i, j].plot(x, x ** (i + j))
        axs[i, j].set_title(f"i, j = {i}, {j}")
```

### 2. Compartir ejes

```python
# crear un arreglo x (variables independientes)
x = np.linspace(-5, 5, 100)
# crear grilla
fig, axs = plt.subplots(2, 3, sharex=True)
fig.subplots_adjust(hspace=0.5, wspace=0.5) 
# iterar sobre la grilla para generar graficos
for i in range(2):
    for j in range(3):
        axs[i, j].plot(x, x ** (i + j))
        axs[i, j].set_title(f"i, j = {i}, {j}")
```

```python
# crear un arreglo x (variables independientes)
x = np.linspace(-5, 5, 100)
# crear grilla
fig, axs = plt.subplots(2, 3, sharex=True, sharey=True)
fig.subplots_adjust(hspace=0.5, wspace=0.5) 
# iterar sobre la grilla para generar graficos
for i in range(2):
    for j in range(3):
        axs[i, j].plot(x, x ** (i + j))
        axs[i, j].set_title(f"i, j = {i}, {j}")
```

### 3. Grillas complejas

```python
# crear la grilla con:
#       - la primera columna divida en 2 axes
#       - la segunda columna divida en 1 axes

fig, axs = plt.subplot_mosaic(
    [
        ['upleft', 'right'], # fila 1
        ['lowleft', 'right'] # fila 2
    ]
)

axs['upleft'].set_title('UPLEFT')
axs['lowleft'].set_title('LOWLEFT')
axs['right'].set_title('RIGHT')
```

```python
# crear la grilla con:
#       - la primera columna divida en 2 axes
#       - la segunda columna divida en 1 axes

fig, axs = plt.subplot_mosaic(
    [
        ['upleft', 'right'], # fila 1
        ['lowleft', 'right'] # fila 2
    ], layout='constrained'
)

axs['upleft'].set_title('UPLEFT')
axs['lowleft'].set_title('LOWLEFT')
axs['right'].set_title('RIGHT')
```

```python
# crear la grilla con:
#       - la primera columna divida en 2 axes
#       - la segunda columna divida en 1 axes

fig, axs = plt.subplot_mosaic(
    [
        ['upleft', 'right'], # fila 1
        ['lowleft', 'right'] # fila 2
    ], layout='constrained'
)

axs['upleft'].set_title('UPLEFT')
axs['lowleft'].set_title('LOWLEFT')
axs['right'].set_title('RIGHT')

axs['upleft'].plot(x, x, 'r')
axs['lowleft'].plot(x, x ** 2, 'c--')
axs['right'].plot(x, x ** 3, 'b-.')
```
