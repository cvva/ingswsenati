### Generar una grafica

crear una grafica sencilla: la de una funcion coseno. Para ello importaremos numpy

```python
import numpy as np

x = np.linspace(0, 2*np.pi, 100)  # variable independiente de 0 a 2*pi (100 datos)
y = np.cos(x) # variable dependiente. el coseno x
```

```python
x
```

```python
y
```

```python
# imprimir el plot
plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, color='red')
plt.show()
```

```python
plt.plot(x, y, color='peru')
plt.show()
```

```python
plt.plot(x, y, color='peru', linestyle='dashed')
plt.show()
```

```python
plt.plot(x, y, 'm--')
plt.show()
```

```python
help(plt.plot)
```

### Exportando la grafica a un archivo de imagen

```python
plt.plot(x, y, 'peru')
plt.savefig('../img/grafico_coseno.png')
```
