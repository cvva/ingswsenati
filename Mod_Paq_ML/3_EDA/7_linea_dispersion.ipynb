# Grafica de Dispersion

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ciudades = pd.read_csv('../data/california_cities.csv')
ciudades
```

```python
ciudades.info()
```

```python
ciudades.describe()
```

```python
ciudades[['latd', 'longd', 'population_total', 'area_total_km2']]
```

```python
# latd  -> latitud
# longd -> longitud
# population_total  -> poblacion
# area_total_km2    -> area total en m2
latitud = ciudades['latd']
longitud = ciudades['longd']
poblacion = ciudades['population_total']
area = ciudades['area_total_km2']
```

```python
fig, ax = plt.subplots()
ax.scatter(longitud, latitud);
```

```python
plt.style.use('Solarize_Light2')

fig, ax = plt.subplots()
ax.scatter(longitud, latitud);
```

```python
fig, ax = plt.subplots()
ax.scatter(longitud, latitud)
ax.set_xlabel('Longitud')
ax.set_ylabel('Latitud');
```

```python
fig, ax = plt.subplots()
ax.scatter(longitud, latitud, s=area)
ax.set_xlabel('Longitud')
ax.set_ylabel('Latitud');
```

```python
fig, ax = plt.subplots()
ax.scatter(longitud, latitud, s=area, c=poblacion, cmap='viridis')
ax.set_xlabel('Longitud')
ax.set_ylabel('Latitud');
```

```python
fig, ax = plt.subplots()
grafico = ax.scatter(longitud, latitud, s=area, c=poblacion, cmap='viridis')
ax.set_xlabel('Longitud')
ax.set_ylabel('Latitud')
fig.colorbar(grafico, label='poblacion');
```

```python
fig, ax = plt.subplots()
grafico = ax.scatter(longitud, latitud, s=area, c=np.log10(poblacion), cmap='viridis')
ax.set_xlabel('Longitud')
ax.set_ylabel('Latitud')
fig.colorbar(grafico, label='$log_{10} poblacion$');
```

```python
fig, ax = plt.subplots()
grafico = ax.scatter(longitud, latitud, s=area, c=np.log10(poblacion), cmap='viridis',
                    vmin=2, vmax=7, alpha=0.5)
ax.set_xlabel('Longitud')
ax.set_ylabel('Latitud')
fig.colorbar(grafico, label='$log_{10} poblacion$');
```

```python
fig, ax = plt.subplots()
grafico = ax.scatter(longitud, latitud, s=area, c=np.log10(poblacion), cmap='viridis',
                    vmin=2, vmax=7, alpha=0.5)
ax.set_xlabel('Longitud')
ax.set_ylabel('Latitud')
fig.colorbar(grafico, label='$log_{10} poblacion$')

for areas in [50, 250, 450]:
    ax.scatter([], [], s=areas, c='k', alpha=0.5, label=str(areas) + '$km^2$')

ax.legend(title='Areas', labelspacing=1)
ax.set_title("Area y Poblacion de diferentes ciudades de California");
```
