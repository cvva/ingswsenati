# Grafico de Histograma

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/histograma_dataset.csv')
df
```

```python
# df_fem = df[df['Sexo']=='Femenino']
df_fem = df.loc[df['Sexo']=='Femenino']
df_fem
```

```python
df_mas = df.loc[df['Sexo']=='Masculino']
df_mas
```

```python
# seleccionar estilo
plt.style.use('seaborn-v0_8-white')
# generar un hitograma basico (conteo de ocurrencia)
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].hist(df_fem['Altura (cm)'])
ax[1].hist(df_mas['Altura (cm)'])
# agregar titulos, etiquetas a los ejes
ax[0].set_title("Altura de mujeres")
ax[1].set_title("Altura de hombres")
ax[0].set_xlabel('Altura (cm)')
ax[1].set_xlabel('Altura (cm)')
ax[0].set_ylabel('Conteo')
plt.show()
```

```python
# seleccionar estilo
plt.style.use('seaborn-v0_8-white')
# generar un hitograma basico (conteo de ocurrencia)
fig, ax = plt.subplots(1, 2, figsize=(10, 4), sharey=True)
ax[0].hist(df_fem['Altura (cm)'], bins=25)
ax[1].hist(df_mas['Altura (cm)'], bins=25)
# agregar titulos, etiquetas a los ejes
ax[0].set_title("Altura de mujeres")
ax[1].set_title("Altura de hombres")
ax[0].set_xlabel('Altura (cm)')
ax[1].set_xlabel('Altura (cm)')
ax[0].set_ylabel('Conteo')
plt.show()
```

```python
# seleccionar estilo
plt.style.use('seaborn-v0_8-white')
# generar un hitograma basico (conteo de ocurrencia)
fig, ax = plt.subplots(1, 2, figsize=(10, 4), sharey=True)
ax[0].hist(df_fem['Altura (cm)'], bins=25, alpha=0.5)
ax[1].hist(df_mas['Altura (cm)'], bins=25, alpha=0.5)
# agregar titulos, etiquetas a los ejes
ax[0].set_title("Altura de mujeres")
ax[1].set_title("Altura de hombres")
ax[0].set_xlabel('Altura (cm)')
ax[1].set_xlabel('Altura (cm)')
ax[0].set_ylabel('Conteo')
# agregar grilla
ax[0].grid()
ax[1].grid()
plt.show()
```

```python
# seleccionar estilo
plt.style.use('seaborn-v0_8-white')
# generar un hitograma basico (conteo de ocurrencia)
fig, ax = plt.subplots(1, 2, figsize=(10, 4), sharey=True)
ax[0].hist(df_fem['Altura (cm)'], bins=25, alpha=0.5, density=True)
ax[1].hist(df_mas['Altura (cm)'], bins=25, alpha=0.5, density=True)
# agregar titulos, etiquetas a los ejes
ax[0].set_title("Altura de mujeres")
ax[1].set_title("Altura de hombres")
ax[0].set_xlabel('Altura (cm)')
ax[1].set_xlabel('Altura (cm)')
ax[0].set_ylabel('Conteo')
# agregar grilla
ax[0].grid()
ax[1].grid()
plt.show()
```

```python
# seleccionar estilo
plt.style.use('seaborn-v0_8-white')
# generar un hitograma basico (conteo de ocurrencia)
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(df_fem['Altura (cm)'], bins=25, alpha=0.5, density=True, label='Muejeres')
ax.hist(df_mas['Altura (cm)'], bins=25, alpha=0.5, density=True, label='Varones')
# agregar titulos, etiquetas a los ejes
ax.set_title("Distribucion de Alturas de Mujeres y Varones")
ax.set_xlabel('Altura (cm)')
ax.set_ylabel('Densidad de Probabilidad')
# agregar grilla
ax.grid()
ax.legend()
plt.show()
```
