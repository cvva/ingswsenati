# Personalizar los elementos de una grafica

- Añadir un titulo a la grafica y etiquetas a los ejes
- Añadir expresiones matematicas al titulo, o a los ejes
- Cambiar los ticks y añadir grillas
- Añadir leyenda a la grafica
- peronalizar elementos graficados, la figura y los ejes
- Personalizacion a traves de `rcParams`

### 1. Añadir un titulo a la grafica y etiquetas a los ejes

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-2, 2, 100)
y = x ** 2 + 5

fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

```python
# figure y axes
fig, ax = plt.subplots()
# personalizar
ax.set_title("Funcion Cuadratica")
ax.set_xlabel("Variables Indepedientes")
ax.set_ylabel("y = x ^ 2 + 5")

ax.plot(x, y)
plt.show()
```

### 2. Añadir expresiones matematicas al titulo o a las etiquetas de los ejes

```python
# figure y axes
fig, ax = plt.subplots()
# personalizar
ax.set_title("Funcion Cuadratica")
ax.set_xlabel("Variables Indepedientes")
ax.set_ylabel("$y = x^2 + 5$") # expresion matematica

ax.plot(x, y)
plt.show()
```

### 3. Cambiar los ticks y añadir grillas

```python
# figure y axes
fig, ax = plt.subplots()
# personalizar
ax.set_title("Funcion Cuadratica")
ax.set_xlabel("Variables Indepedientes")
ax.set_ylabel("$y = x^2 + 5$") # expresion matematica

ax.set_xticks([-2, -1, 0, 1, 2]) # ticks eje x
ax.set_yticks([5, 6, 7, 8, 9]) # ticks eje y
ax.grid(True) # grilla para ambos ejex

ax.plot(x, y)
plt.show()
```

```python
# figure y axes
fig, ax = plt.subplots()
# personalizar
ax.set_title("Funcion Cuadratica")
ax.set_xlabel("Variables Indepedientes")
ax.set_ylabel("$y = x^2 + 5$") # expresion matematica

ax.set_xticks([-2, -1, 0, 1, 2]) # ticks eje x
ax.set_yticks([5, 6, 7, 8, 9]) # ticks eje y
ax.grid(True, axis='y') # grilla para ambos ejex

ax.plot(x, y)
plt.show()
```

```python
# figure y axes
fig, ax = plt.subplots()
# personalizar
ax.set_title("Funcion Cuadratica")
ax.set_xlabel("Variables Indepedientes")
ax.set_ylabel("$y = x^2 + 5$") # expresion matematica

ax.set_xticks([-2, -1, 0, 1, 2]) # ticks eje x
ax.set_yticks([5, 6, 7, 8, 9]) # ticks eje y
ax.grid(True, linestyle='-.', linewidth=1) # grilla para ambos ejex

ax.plot(x, y)
plt.show()
```

### 4. Añadir leyenda a la grafica

```python
x = np.linspace(-2, 2, 100)
y1 = 5 * x + 2
y2 = 3 * x ** 2
y3 = 4 * x ** 3

fig, ax = plt.subplots()
ax.set_title('Funcion Lineal, Cuadratica, cubica')
ax.set_xlabel('X')
ax.set_ylabel('Y')

ax.set_xticks([-2, -1, 0, 1, 2])
ax.grid(True, linestyle='--', linewidth=1)

ax.plot(x, y1, label='$5x+2$')
ax.plot(x, y2, label='$3x^2$')
ax.plot(x, y3, label='$4x^3$')

ax.legend()
plt.show()
```
