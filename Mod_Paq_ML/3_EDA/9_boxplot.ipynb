# ¿Que es un BoxPlot?

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/histograma_dataset.csv')

df_fem = df.loc[df['Sexo']=='Femenino']
df_mas = df.loc[df['Sexo']=='Masculino']
```

```python
df_fem 
```

```python
df_mas
```

```python
# Boxplot Basico
fig, ax = plt.subplots(figsize=(5,4))
ax.boxplot(df_fem['Altura (cm)'])

ax.set_title('Distribucion de altura de mujeres')
ax.set_xlabel('Sexo')
ax.set_ylabel('Altura (cm)')

ax.grid(True)
plt.show()
```

### Personalizar

```python
### Personalizar
fig, ax = plt.subplots(figsize=(5,4))
ax.boxplot([df_fem['Altura (cm)'], df_mas['Altura (cm)']], tick_labels=['Mujeres', 'Varones'])

ax.set_title('Distribucion de Altura Mujeres y Varones')
ax.set_ylabel('Altura (cm)')
ax.set_xlabel('Sexo')

ax.grid()
plt.show()
```

```python
### Personalizar
fig, ax = plt.subplots(figsize=(5,4))
bx_dic = ax.boxplot([df_fem['Altura (cm)'], df_mas['Altura (cm)']], tick_labels=['Mujeres', 'Varones'],
            medianprops={'color':'red'}, patch_artist=True)

colors = ['lightblue', 'lightgreen']
for patch, color in zip(bx_dic['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)

ax.set_title('Distribucion de Altura Mujeres y Varones')
ax.set_ylabel('Altura (cm)')
ax.set_xlabel('Sexo')

ax.grid()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5,4))
bx_dic = ax.boxplot([df_fem['Altura (cm)'], df_mas['Altura (cm)']], tick_labels=['Mujeres', 'Varones'],
            medianprops={'color':'red'}, patch_artist=True, vert=False)

colors = ['lightblue', 'lightgreen']
for patch, color in zip(bx_dic['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)

ax.set_title('Distribucion de Altura Mujeres y Varones')
ax.set_ylabel('Sexo')
ax.set_xlabel('Altura (cm)')

ax.grid()
plt.show()
```
