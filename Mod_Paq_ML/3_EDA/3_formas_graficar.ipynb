# Formas de generar graficas

### Forma 1: estilo stateful

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10*np.pi, 400)
y = np.sin(x)

plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y)
# crear una variable figure
figure = plt.gcf()
type(figure)
```

```python
plt.plot(x, y)
# crear una variable figure
axes = plt.gca()
type(axes)
```

### Forma 2: al estilo stateless

```python
fig, ax = plt.subplots()

ax.plot(x, y)
plt.show()
```

### Enfoque stateful (forma 1) o stateless (forma 2)

```python
# usando stateful
plt.plot(x , y)
# extraer el puntero de la figura (gcf) y cambiar el tamaño
plt.gcf().set_size_inches(10, 8)
# extraer el puntero del axes (gca) y fijar sus ticks
plt.gca().set_xticks([0, 15, 30])

plt.show()
```

```python
# usando stateless

# generar la figura y axes vacios
fig, ax = plt.subplots()
# ajustar el tamaño de la figura
fig.set_size_inches(10, 8)
# fijar los ticks a los axes
ax.set_xticks([0, 15 ,30])
ax.set_yticks([-1, 0, 1])

#graficar los datos
ax.plot(x, y)
plt.show()
```

```python
# usando stateful
plt.plot(x , y)
plt.plot(x, 2*y)
# extraer el puntero de la figura (gcf) y cambiar el tamaño
plt.gcf().set_size_inches(10, 5)
# extraer el puntero del axes (gca) y fijar sus ticks
plt.gca().set_xticks([0, 15, 30])

plt.show()
```

```python
# usando stateless

# generar la figura y axes vacios
fig1, ax1 = plt.subplots()
fig2, ax2 = plt.subplots()
# ajustar el tamaño de la figura
fig1.set_size_inches(10, 5)
fig2.set_size_inches(5, 4)
# fijar los ticks a los axes
ax1.set_xticks([0, 15 ,30])
ax2.set_xticks([0, 10, 20, 30])

#graficar los datos
ax1.plot(x, y)
ax2.plot(x, 2*y)
plt.show()
```
