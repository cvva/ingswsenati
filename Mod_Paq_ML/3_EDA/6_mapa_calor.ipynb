# Visualizacion de mapas de calor

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
matriz = np.array(
    [
        [1.0, -0.8, 0.8, 0.1],
        [-0.8, 1.0, 0.7, 0.6],
        [0.8, 0.7, 1, -0.1],
        [0.1, 0.6, -0.1, 1]
    ]
)
```

```python
matriz
```

```python
fig, axs = plt.subplots()
axs.matshow(matriz)
plt.show()
```

```python
fig, axs = plt.subplots()
axs.matshow(matriz, cmap='coolwarm')
plt.show()
```

```python
fig, axs = plt.subplots()
mapa = axs.matshow(matriz, cmap='coolwarm', vmin=-1, vmax=1)
fig.colorbar(mapa)
positions = [0, 1, 2, 3]
labels = ['Peso', 'Altura', 'Edad', 'Sueldo']
axs.set_xticks(positions, labels)
axs.set_yticks(positions, labels)
plt.show()
```
