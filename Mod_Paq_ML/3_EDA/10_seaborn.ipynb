# Seaborn

```python
import seaborn as sns
```

```python
tip = sns.load_dataset('tips')
tip
```

```python
tip.info()
```

```python
tip.describe()
```

### 1. Distribuciones: Graficas que muestran la distribucion en una, dos o mas variables

##### Distribucion Univariados

```python
sns.displot(tip, x='total_bill');
```

```python
sns.rugplot(tip, x='total_bill')
```

```python
sns.displot(tip, x='total_bill', kde=True);
```

```python
sns.kdeplot(data=tip, x='total_bill');
```

```python
sns.kdeplot(data=tip, x='total_bill', fill=True);
```

```python
sns.kdeplot(data=tip, x='total_bill', bw_adjust=500);
```

##### Distribucion Bivariado

```python
sns.jointplot(data=tip, x='total_bill', y='tip', kind='scatter');
```

```python
sns.jointplot(data=tip, x='total_bill', y='tip', kind='hex');
```

```python
sns.jointplot(data=tip, x='total_bill', y='tip', kind='kde');
```

```python
sns.kdeplot(data=tip, x='total_bill', y='tip');
```

##### Distribucion de Varias Variables

```python
tip
```

```python
sns.pairplot(tip);
```

```python
sns.relplot(data=tip, x='total_bill', y='tip', hue='sex');
```

```python
sns.relplot(data=tip, x='total_bill', y='tip', hue='size');
```

```python
sns.relplot(data=tip, x='total_bill', y='tip', hue='sex', style='time');
```

```python
sns.relplot(data=tip, x='total_bill', y='tip', size='size');
```

```python
sns.scatterplot(data=tip, x='total_bill', y='tip', size='size');
```

```python
sns.relplot(data=tip,
            x='total_bill',
            y='tip',
            hue='sex',
            style='time',
            col='time');
```

```python
sns.relplot(data=tip,
            x='total_bill',
            y='tip',
            hue='sex',
            row='smoker');
```

```python
sns.relplot(data=tip, x='total_bill', y='tip',
            hue='sex', row='smoker', col='time', height=3.5);
```
