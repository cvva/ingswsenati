# Limpieza y Analisis Exploratorio de Datos

En el año 1912, el transatlántico RMS Titanic sufrió uno de los naufragios más conocidos de la historia, provocando la muerte de una gran cantidad de pasajeros y tripulantes. A pesar de contar con tecnología avanzada para su época, no todos los pasajeros tuvieron las mismas probabilidades de sobrevivir. Factores como el sexo, la edad, la clase social, el precio del pasaje y la ubicación de embarque podrían haber influido significativamente en la supervivencia.

El dataset Titanic contiene información demográfica y socioeconómica de los pasajeros, pero presenta problemas de calidad de datos, tales como valores faltantes, variables mal tipificadas y columnas irrelevantes o con información no estructurada. Estas deficiencias impiden un análisis confiable si no se realiza previamente un adecuado proceso de limpieza y preparación de datos.

Ante esta situación, surge la necesidad de aplicar técnicas de limpieza de datos y análisis exploratorio (EDA) utilizando el lenguaje de programación Python y librerías especializadas como Pandas, NumPy, Matplotlib y Seaborn, con el fin de identificar patrones, relaciones y factores determinantes que influyeron en la supervivencia de los pasajeros del Titanic.

## Planteamiento del problema ❓

¿Qué variables demográficas y socioeconómicas influyeron de manera significativa en la probabilidad de supervivencia de los pasajeros del Titanic?

## 🎯 Objetivo General

Analizar y limpiar el dataset Titanic para identificar los principales factores que influyeron en la supervivencia de los pasajeros mediante técnicas de análisis exploratorio de datos utilizando Python.

## 📊 Descripción de Variables – Dataset Titanic

| Variable | Tipo de Dato | Descripción | Valores / Observaciones |
|---------|--------------|-------------|-------------------------|
| PassengerId | Numérico (int) | Identificador único del pasajero | Valor entero incremental |
| Survived | Categórico (binario) | Indica si el pasajero sobrevivió | 0 = No, 1 = Sí |
| Pclass | Categórico (ordinal) | Clase del pasajero | 1 = Primera, 2 = Segunda, 3 = Tercera |
| Name | Texto (string) | Nombre completo del pasajero | Contiene títulos (Mr, Mrs, Miss, etc.) |
| Sex | Categórico | Sexo del pasajero | male, female |
| Age | Numérico (float) | Edad del pasajero en años | Contiene valores nulos |
| SibSp | Numérico (int) | Número de hermanos/esposos a bordo | Valores enteros |
| Parch | Numérico (int) | Número de padres/hijos a bordo | Valores enteros |
| Ticket | Texto (string) | Número de boleto | Formato alfanumérico |
| Fare | Numérico (float) | Tarifa pagada por el pasaje | Puede contener outliers |
| Cabin | Texto (string) | Número de cabina | Alta cantidad de valores nulos |
| Embarked | Categórico | Puerto de embarque | C = Cherbourg, Q = Queenstown, S = Southampton |
