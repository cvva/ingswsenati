# Ejemplo de Regresion Lineal Simple

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# crear datos de ejemplos con numpy
x = np.array([1.2, 3, 3.2, 2.5, 5, 6, 4, 8])
y = np.array([2, 3, 3.4, 3.1, 4, 4.7, 3.8, 7])
```

```python
# graficar
fig, ax = plt.subplots()
ax.scatter(x, y, color='red')
ax.set_title('Datos Ejemplo')
ax.set_xlabel('Variable X')
ax.set_ylabel('Variable Y')
plt.show()
```

```python
import random

w1 = 1
w0 = 2
valid_err = 1.5
err_ant = 1000
best = [w1, w0]

def f(x):
    return w1 * x + w0

def error(y_pred):
    return np.sum((y - y_pred)**2)

n = 100
for i in range(n):
    y_pred = f(x)
    err = error(y_pred)
    print(f"w0: {round(w0, 2)} | w1: {round(w1, 2)} | error: {round(err, 2)}")

    if err < err_ant:
        best = [w1, w0]
    else:
        w1 = best[0] + random.uniform(-0.1, 0.1)
        w = best[1] + random.uniform(-0.1, 0.1)

    err_ant = err

    if err < valid_err:
        print(f"r: {round(w1, 2)}, {round(w0, 2)}")
        break

w1, w0 = best
y_pred = f(x)
err = error(y_pred)
print(f"w0: {round(w0, 2)} | w1: {round(w1, 2)} | error: {round(err, 2)}")
```

```python
x0 = x.copy()
x0.sort()
y0 = f(x0)
plt.scatter(x, y, color='green')
plt.plot(x0, y0, color='red')
plt.show()
```

```python
from sklearn.linear_model import  LinearRegression
```

```python
modelo = LinearRegression()

modelo.fit(x.reshape(-1, 1), y)

x_hat = np.linspace(1.2, 8, 100)
y_hat = modelo.predict(x_hat.reshape(-1, 1))

plt.scatter(x, y)
plt.plot(x_hat, y_hat, color='red')
plt.ylabel('Ingresos')
plt.xlabel('Años de Experiencia')
plt.legend(['Datos reales', 'Modelo RL'])
plt.show()
```
