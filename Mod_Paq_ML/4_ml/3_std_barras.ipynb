# Grafico de Barras

```python
import pandas as pd

df = pd.read_csv('../data/data-prestamos.csv')
df
```

```python
import seaborn as sns

sns.countplot(data=df, x='nivel');
```

```python
import matplotlib.pyplot as plt

sns.countplot(data=df, x='estado')
plt.xticks(rotation=45);
```

```python
sns.countplot(data=df, x='nivel', stat='percent');
```

```python
sns.countplot(data=df, x='nivel', stat='proportion');
```

```python
sns.countplot(data=df, x='estado', stat='percent')
plt.xticks(rotation=45);
```

```python
df['nivel'].value_counts()
```

```python
df['nivel'].value_counts().index
```

```python
sns.countplot(data=df, x='nivel', order=df['nivel'].value_counts().index, stat='percent');
```

```python
sns.countplot(data=df, x='estado', order=df['estado'].value_counts().index, stat='percent')
plt.xticks(rotation=45);
```

```python
sns.countplot(data=df, y='nivel', order=df['nivel'].value_counts().index, stat='percent');
```

```python
sns.countplot(data=df, y='estado', order=df['estado'].value_counts().index, stat='percent');
```
