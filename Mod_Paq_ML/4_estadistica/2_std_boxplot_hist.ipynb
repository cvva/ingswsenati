# Grafica de Caja (boxplot)

```python
import pandas as pd

df = pd.read_csv('../data/data-boxplots.csv')
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

ax = sns.boxplot(data=df, y='var1')
ax.set_title('Boxplot Var1')
ax.set_ylabel('Valores')
ax.set_xlabel('Var1')
plt.show()
```

```python
ax = sns.boxplot(data=df, y='var2')
ax.set_title('Boxplot Var2')
ax.set_ylabel('Valores')
ax.set_xlabel('Var2')
plt.show()
```

```python
ax = sns.boxplot(data=df, y='var2')
ax = sns.stripplot(data=df, y='var2', jitter=True, alpha=0.5)
ax.set_title('Boxplot Var2')
ax.set_ylabel('Valores')
ax.set_xlabel('Var2')
plt.show()
```

```python
ax = sns.boxplot(data=df, y='var2', showfliers=False)
ax = sns.stripplot(data=df, y='var2', jitter=True, alpha=0.5)
ax.set_title('Boxplot Var2')
ax.set_ylabel('Valores')
ax.set_xlabel('Var2')                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                  
plt.show()
```

```python
ax = sns.boxplot(data=df, y='var2', showfliers=False)
ax = sns.stripplot(data=df, y='var2', color='red', jitter=True, alpha=0.5)
ax.set_title('Boxplot Var2')
ax.set_ylabel('Valores')
ax.set_xlabel('Var2')                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                  
plt.show()
```

# Histograma

```python
df = pd.read_csv('../data/data-histogramas.csv')
df
```

```python
ax = sns.histplot(data=df, x='var_1')
ax.set_title('Histograma de var_1')
ax.set_xlabel('Variables')
ax.set_ylabel('Frecuencian (conteo)')
plt.show()
```

```python
ax = sns.histplot(data=df, x='var_2')
ax.set_title('Histograma de var_2')
ax.set_xlabel('Variables')
ax.set_ylabel('Frecuencian (conteo)')
plt.show()
```

```python
ax = sns.histplot(data=df, x='var_3')
ax.set_title('Histograma de var_3')
ax.set_xlabel('Variables')
ax.set_ylabel('Frecuencian (conteo)')
plt.show()
```

```python
ax = sns.histplot(data=df, x='var_1', bins=100)
ax.set_title('Histograma de var_1')
ax.set_xlabel('Variables')
ax.set_ylabel('Frecuencian (conteo)')
plt.show()
```

```python
ax = sns.histplot(data=df, x='var_1', bins=50)
ax.set_title('Histograma de var_1')
ax.set_xlabel('Variables')
ax.set_ylabel('Frecuencian (conteo)')
# calcular y dibujar media y desviacion estandar
media, std = df['var_1'].mean(), df['var_1'].std()
ax.axvline(media, color='red', linestyle='--', linewidth=2, label=f'Media: {media:.2f}')
ax.axvline(media + std, color='orange', linestyle=':', linewidth=2, label=f'Std: {std:.2f}')
ax.axvline(media - std, color='orange', linestyle=':', linewidth=2)
ax.legend()
plt.show()
```

```python
ax = sns.histplot(data=df, x='var_1', stat='density')
ax.set_title('Histograma de var_1')
ax.set_xlabel('Variables')
ax.set_ylabel('Densidad de probabilidad')
plt.show()
```

```python
ax = sns.histplot(data=df, x='var_1', stat='density', kde=True)
ax.set_title('Histograma de var_1')
ax.set_xlabel('Variables')
ax.set_ylabel('Densidad de probabilidad')
plt.show()
```

```python
ax = sns.histplot(data=df, x='var_1', stat='density', kde=True, bins=100)
ax.set_title('Histograma de var_1')
ax.set_xlabel('Variables')
ax.set_ylabel('Densidad de probabilidad')
plt.show()
```
