# Seleccion y manipulacio de columnas y filas

- Diferentes formas de seleccionar una o multiples columnas de nuestro DataFrame
- Diferentes formas de seleccionar una o multiples filas de nuestro DataFrame
- Como seleccionar porciones de nuestro DataFrame (es decir un rango de filas y columnas)
- Como modificar columnas y filas del DataFrame (cambiar nombre, eliminar, agregar...)

```python
import pandas as pd

df = pd.read_csv('../data/peliculas.csv')
df
```

### 1. Seleccion de columnas

```python
# seleccion de una columna

# fb_likes = df['movie_facebook_likes']
fb_likes = df.movie_facebook_likes
fb_likes
```

```python
type(fb_likes)
```

```python
# seleccion de muliples columnas

df_cols = df[['director_name', 'duration', 'gross']]
df_cols
```

Tambien podemos hacer uso de la propiedad `loc`

```python
df.loc[:,['duration', 'gross']]
```

El metodo `select_dtype` permite seleccionar columnas con base en el tipo de dato

```python
df.dtypes.value_counts()
```

```python
# seleccionar solo las columas de tipo "int64"

# df_int64 = df.select_dtypes(include='int64')
df_int64 = df.select_dtypes(include=['int64'])
df_int64
```

```python
# seleccionar solo las columas de tipo "float64"

df_float64 = df.select_dtypes(include='float64')
df_float64
```

### 2. Seleccion de filas

En este caso disponemos de los metodos `iloc` (seleccion en base en el indice) y `loc` (usado anteriormente)

```python
# seleccionar una fila con nase en un indice (iloc)
row_3 = df.iloc[3]
row_3
```

```python
# seleccion de un afila, pandas retorna una serie
type(row_3)
```

```python
# seleccionar filas de la 3 a la 6
row_3_6 = df.iloc[3:7]
row_3_6
```

```python
# seleccionar solo las filas para las cuales la columna "director_name" es igual a "Steven Spielberg"

df_ss = df.loc[df['director_name']=='Steven Spielberg']
df_ss
```

```python
# seleccionar solo las filas para las cuales la columna "director_name"
# es igual a "Steven Spielberg" o "Quentin Tarantino"
df_ss_qt = df.loc[(df['director_name']=='Steven Spielberg') | (df['director_name']=='Quentin Tarantino')] 
df_ss_qt
```

```python
# seleccionar solo las filas para las cuales la columna "director_name"
# es igual a "Steven Spielberg" y "imdb_score" es mayor a 8.5

df_ss_imdb = df.loc[(df['director_name']=='Steven Spielberg') & (df['imdb_score'] > 8.5)] 
df_ss_imdb
```

### 3. Seleccion de filas y columnas

podemos combinar los metodos vistos anteriormente para seleccionar porciones (filas, columnas) de un DataFrame
Las sintaxis a usar denpenden de si recurrimos al metodo `iloc` o al medoto `loc`:

- Con el metodo `iloc`. En este caso la sintaxis es -> df.iloc[idxs_filas, idxs_col]
- Con el metodo `loc`. En este caso la sintaxis es -> df.loc[nombre_fila, nombre_col]

```python
# ejm 1: selecionar de las filas 3 a la 5 y de las columnas 2 a la 5

df_ejem1 = df.iloc[3:6, 2:6]
df_ejem1
```

```python
# ejem 2: seleccionar la misma porcion anterior (ejem 1) con loc
df_ejem2 = df.loc[3:5, 'num_critic_for_reviews':'actor_3_facebook_likes']
df_ejem2
```

```python
row_names = [3,4,5]
col_name = ['num_critic_for_reviews','duration','director_facebook_likes','actor_3_facebook_likes']
df_ejem2 = df.loc[row_names, col_name]
df_ejem2
```
