# Metodos basicos y operacones sobre un DataFrame

- Metodos que nos permiten obtener informacion general de un dataframe
- metodos para realizar operaciones matematicas y de sintesis (sumarizacion) de un dataframe
- el metodo `apply` que nos permite aplicar cualquier tipo de funcion al dataframe

### 1. Metodos que nos permiten obtener informacion general de un dataframe

```python
import pandas as pd

df = pd.read_csv('../data/peliculas.csv')
df
```

El metodo `info()` nos permite imprimir en pantalla ifnormacion general sobre la cantidad de datos "nulos - no nulos" y el tipo de dato presentes en cada columna

```python
df.info()
```

Los metodos `head()`, `tail()`, `sample()` nos permitir imprimir en pantalla las primeras filas del DataFrame, las ultimas filas o una seleccion aleatoria de filas respectivamente

```python
df.head()
```

```python
df.head(10)
```

```python
df.tail()
```

```python
df.tail(10)
```

```python
df.sample()
```

```python
df.sample(10)
```

El metodo `describe()` permite imprimir informacion estadistica de las columnas que contienen variables numericas

```python
df.describe()
```

El metodo `value_counts()` tambien resulta util pues permite imprimir en pantalla el conteo de valores presentes en cada columna, funciona tanto para variables numericas como categoricas

```python
df['duration'].value_counts()
```

```python
df['color'].value_counts()
```

### 2. Operacones matematicas y de sintesis

- `min()` y `max()`: calcula el maximo / minimo de cada fila o columna

- `mean()`, `std()`, `median()`: calcula la media/desviacion/mediana de cada fila o columna

```python
df.min()
```

```python
df['num_critic_for_reviews'].min()
```

### 3. Apply: aplicar una funcion especifica por filas o columas

```python
import numpy as np

df.select_dtypes(include='number').apply(np.sqrt)
```

```python
df['duration'].apply(np.sqrt)
```

```python
df[['duration','gross']].apply(np.sqrt)
```

```python
df['genres']
```

```python
def change_str(string):
    if "|" in string:
        return string.replace("|", ", ")
```

```python
df['genres'].apply(change_str)
```
