### 2. Principales Atributos de un DataFrame

```python
# numeros de filas y columnas de un dataframe
df.shape
```

```python
print(f"Numero de filas: {df.shape[0]}")  # numero de filas
print(f"Numero de columnas: {df.shape[1]}") # numero de columnas
```

```python
indice = df.index
print(indice)
print(type(indice))
```

```python
columnas = df.columns
print(columnas)
print(type(columnas))
```

```python
datos = df.values
print(datos)
print(type(datos))
```
