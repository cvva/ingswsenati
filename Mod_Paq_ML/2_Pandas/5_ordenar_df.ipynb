# Formas de ordenar un DataFrame

- Orgamizar las columnas del DataFrame
- Organizar las filas de DataFrame con base en la informacion contenida en una o multiples columnas
- Organizar las filas con base en el indice del DataFrame

### 1. Organizar las columnas del DataFrame

```python
import pandas as pd

df = pd.read_csv('../data/peliculas.csv')
df
```

```python
df.dtypes
```

```python
df.info()
```

```python
# agrupar las columnas en categoricas y numericas

# columnas categoricas
cols_cat = ['color', 'director_name', 'actor_2_name', 'genres', 'actor_1_name', 'movie_title', 
            'actor_3_name', 'plot_keywords', 'movie_imdb_link', 'language']
# columnas numericas
cols_num = ['num_critic_for_reviews','duration', 'director_facebook_likes', 'actor_3_facebook_likes',
 'actor_1_facebook_likes', 'gross', 'num_voted_users', 'cast_total_facebook_likes', 'facenumber_in_poster', 
 'num_user_for_reviews', 'budget', 'title_year', 'actor_2_facebook_likes', 'imdb_score', 'aspect_ratio', 'movie_facebook_likes']

col_order = cols_cat + cols_num
print(col_order)
```

```python
df = df[col_order]
df
```

### 2. Ordenar las filas con base en los valores de una o multiple columnas

```python
# ordenar el dataframe con base a la columna "movie_title" alfabeticamente
df_sorted_movie = df.sort_values(by='movie_title')
df_sorted_movie
```

```python
# ordenar el dataframe con base a la columna "movie_title" alfabeticamente inverso
df_sorted_movie = df.sort_values(by='movie_title', ascending=False)
df_sorted_movie
```

```python
df_budget_gross = df.sort_values(by=['budget', 'gross'], ascending=False)
df_budget_gross
```

```python
df_budget_gross[['budget','gross']]
```

### 3. Organizas las filas del DataFrame con base en el indice

```python
df_sort_idx = df_budget_gross.sort_index()
df_sort_idx[['gross', 'budget']]
```

```python
df_sort_idx = df_budget_gross.sort_index(ascending=False)
df_sort_idx[['gross', 'budget']]
```
