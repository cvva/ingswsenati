### 2. Union Interna

![](../img/union_interna.png)

> En una union interna (**inner join**), **se conservan unicamente las filas cuyos valores en la columna clave aparecen en ambas tablas**

> Es equivalente a hacer la **interseccion** entre dos tablas usando la columna en comun como criterio de coincidencia

![](../img/union_interna_ejemplo.png)

```python
help(pd.merge)
```

how : {'left', 'right', 'outer', 'inner', 'cross', 'left_anti', 'right_anti}, default 'inner'

```python
# añadir elementos del dataframe "ordenes" donde coincida con el id_cliente
union_interna = pd.merge(clientes, ordenes, on='id_cliente')
union_interna
```

```python
# 
union_interna = pd.merge(clientes, ordenes, on='id_cliente', how='inner')
union_interna
```

### 3. Union Izquierda

> En una union izquierda **se conservan todas las filas de la tabla izquierda y solo las filas coincidentes de la tabla derecha** usando la columna clave como criterio. Las filas sin coincidenia en la derecha se completan con valores nulos

![](../img/union_izquierda.png)

![](../img/union_izquierda_ejemplo.png)

```python
union_izquierda = pd.merge(clientes, ordenes, on='id_cliente', how='left')
union_izquierda
```

### 4. Union Derecha

> En una union derecha **se conservan todas las filas de la tabla derecha y solo las filas coincidentes de la tabla izquierda**, usando la columna clave como criterio. Las filas sin coincidencia en la izquierda se completan con valores nulos.

![](../img/union_derecha.png)

![](../img/union_derecha_ejemplo.png)

```python
union_derecha = pd.merge(clientes, ordenes, on='id_cliente', how='right')
union_derecha
```

```python
union_derecha = pd.merge(ordenes, clientes, on='id_cliente', how='right')
union_derecha
```

### 5. Union Externa

> En una union externa **se conservan todas las filas de ambas tablas, coincidan o no con la columna clave**. Las filas sin coincidencia se compeltan con valores nultos en las colunas de las tablas opuestas

![](../img/union_externa.png)

![](../img/union_externa_ejemplo.png)

```python
union_externa = pd.merge(clientes, ordenes, on='id_cliente', how='outer')
union_externa
```

# Concatenando DataFrames

### 1. Los set de datos

ventas_sucursal_1.csv

ventas_sucursal_2.csv

cantidades.csv

```python
import pandas as pd
from IPython.display import display, HTML

ventas_suc1 = pd.read_csv('../data/ventas_sucursal_1.csv')
ventas_suc2 = pd.read_csv('../data/ventas_sucursal_2.csv')
cants = pd.read_csv('../data/cantidades.csv')

def display_dfs(dfs):
    html = '<div style="display:flex">'
    for df in dfs:
        html += '<div style="margin-right: 25px">'
        html += df.to_html()
        html += '</div>'
    html += '</div>'
    display(HTML(html))

display_dfs([ventas_suc1, ventas_suc2, cants])
```

### 2. Problema a resolver

Objetivo: unir los 3 dataframes de manera tal que aparezcan todos los productos, con sus correspondientes montos y cantidades

### 3. Contatenacion

##### 3.1 Contatenacion Basica

```python
resultado = pd.concat([ventas_suc1, ventas_suc2, cants])

display_dfs([ventas_suc1, ventas_suc2, cants, resultado])
```

##### 3.2 Concatenacion vertical y horizontal

```python
ventas_sucs = pd.concat([ventas_suc1, ventas_suc2])
ventas_sucs
```

```python
ventas_sucs = pd.concat([ventas_suc1, ventas_suc2], ignore_index=True)
ventas_sucs
```

```python
resultado = pd.concat([ventas_sucs, cants], axis=1)
resultado
```

```python
from numpy._core.numeric import  indices
# listados de los dfs originales pero con sus indices modificados
dfs = [ventas_sucs.set_index('ID_producto'), cants.set_index('ID_producto')]
# concatenacion horizontal
resultado = pd.concat(dfs, axis=1).reset_index()
resultado
```
