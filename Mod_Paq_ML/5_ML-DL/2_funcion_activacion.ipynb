## Funciones de Activacion

```python
import numpy as np
import matplotlib.pyplot as plt

N = 1000
x = np.linspace(-5, 5, num=N)
```

```python
x
```

### Funcion lineal

$$y=mx+b$$

```python
def f(x):
    return x

plt.plot(x, f(x))
plt.grid()
plt.show()
```

### Funcion escalon o de Heaviside

$$H(x) =
\begin{cases}
    0, \quad \text{para,} x<0\\
    1, \quad \text{para,} x>=0\\
\end{cases}
$$

```python
def H(x):
    y = np.zeros(len(x))
    for idx, x in enumerate(x):
        if x >= 0:
            y[idx] = 1
    
    return y

# y = H(x)
plt.plot(x, H(x))
plt.grid()
plt.show()
```

### Funcion sigmoide

$$f(x)=\frac{1}{1+e^{-x}}$$

```python
def f(x):
    return 1/(1+np.exp(-x))

plt.plot(x, f(x))
plt.grid()
plt.show()
```

### Funcion tangente hiperbolica

$$f(x)=\frac{2}{1+e^{-2x}}-1$$

```python
def f(x):
    return 2/(1+np.exp(-2*x))-1
    # return np.tanh(x)

plt.plot(x, f(x))
plt.grid()
plt.show()
```

### Funcion RELU

$$R(x)=max(0,x)$$

```python
def f(x):
    return np.maximum(x, 0)

plt.plot(x, f(x))
plt.grid()
plt.show()
```
