# ATRIBUTOS BASICOS DE UN ARREGLO

```python
import numpy as np

a = np.array([[12,25,3],[7,14,92]])
b = np.array([3.4,7.2,9.8])

print(f"Arreglo Bidimensional a:\n {a}")
print(f"Arreglo Unidimensional b: {b}")
```

### Atributo `dtype`

indica el tipo de dato con el que esta almacenado cada elemento del arreglo

```python
a.dtype
```

```python
b.dtype
```

### Atributo `shape`

indica la dimensionalidad del arreglo

```python
a.shape  # (2,3) -> 2 filas y 3 columnas
```

```python
b.shape  # (3,) -> 3 elementos en una sola fila
```

### Atributo `size`

indica el numero total de elementos del arreglo (independientemente de la dimensionalidad)

```python
a.size # 6 elementos en total (2 x 3)
```

```python
b.size # 3 elementos en total
```
