#### FUNCIONES MATEMATICAS

Permiten aplicar funciones matematicas sobre cada elemento de un arreglo. Las principales son: `add`, `subtract`, `multiply`, `divide`, `remainder`, `exp`, `log`, `sqrt`, `power`, `abs` son las mas usandas

```python
import numpy as np

A = np.random.randn(5,4) # matriz 5x4
B = np.random.randn(5,4)
print(A)
print()
print(B)
```

```python
C = np.multiply(A, B) # multiplicacion elemento a elemento
print(C)
```

```python
D = np.power(A, 2) # eleeva al cuadrado cada elemento de la matriz A
print(D)
```

#### FUNCION DEL ALGEBRA LINEAL

- producto punto (`dot`)
- producto de matrices (`matmul`)
- valores propios de una matriz (`eig`)
- norma de una matriz (`norm`)
- inversa de una matriz (`inv`)

```python
matriz1 = np.random.randn(3, 4) # matriz 3x4
matriz2 = np.random.randn(4, 5) # matriz 4 x 5
matriz3 = np.matmul(matriz1, matriz2) # producto matricial 3 x 5

print(f"Matriz 1 de tamaño: {matriz1.shape}")
print(f"Matriz 2 de tamaño: {matriz2.shape}")
print(f"Matriz Resultante:\n {matriz3}")
print(f"de tamaño: {matriz3.shape}")
```

#### FUNCIONES TRIGONOTRICAS

- seno (`sin`)
- coseno (`cos`)
- tangente (`tan`)

```python
# calcular el coseno de cada elemento de una matriz
E = np.random.randn(10, 4)
print(E)
print()
coseno = np.cos(E)
print(coseno)
print(coseno.shape)
```
