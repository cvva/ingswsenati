## COPIAS SHALLOW Y DEEP

#### Copias Shallow

```python
import numpy as np

arr = np.array([5, 27, 82, 3, 14, 5])
print(arr)
```

```python
# al hacer slicing del arreglo anterior y almacenarlo en un nuevo arreglo
slc = arr[3:]
print(slc)
```

```python
# al modificar el contenido del slice (slc) que acabamos de crear
slc[1] = -10
print(slc)
```

```python
# veamos el contenido del arreglo original
print(arr)
```

#### Copias Deep

```python
slc2 = arr[3:].copy()
print(slc)
```

```python
slc2[2] = 999
print(slc2)
```

```python
print(arr)
```
