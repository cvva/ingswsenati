### LISTAS DE PYTHON VS ARREGLOS DE NUMPY

```python
import numpy as np
import time

N = 1_000_000
lista = list(range(N))
arreglo = np.arange(N)

# loop python
t10 = time.time()
new_list = []
for x in lista:
    new_list.append(x * 2)
t11 = time.time()

# Numpy
t2 = time.time()
resultado_numpy = arreglo * 2
t3 = time.time()

print(f"Python loop :   {(t11-t10)*1000:.2f} ms")
print(f"Numpy       :   {(t3-t2)*1000:.2f} ms")
```
