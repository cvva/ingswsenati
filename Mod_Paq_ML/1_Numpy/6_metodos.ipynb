## METODOS DE UN ARREGLO

Un arreglo de Numpy es un objeto tipo `ndarray` en Python. Y por ser un objeto tiene asociado unos metodos que permite ejecutar diferentes operaciones sobre el arreglo.

```python
import numpy as np

vector = np.arange(5, 20)
matriz = np.linspace(1, 10, 20).reshape(5,4)

print("Vector:\n", vector)
print("Matriz:\n", matriz)
```

#### METODOS `max()`, `min()`, `argmax()`, `argmin()`

```python
vector_maximo = vector.max()
vector_minimo = vector.min()
print(f"Vector: {vector}")
print(f"Valor maximo del vector: {vector_maximo}")
print(f"Valor minimo del vector: {vector_minimo}")
```

```python
matriz_maximo = matriz.max()
matriz_minimo = matriz.min()
print(f"matriz: {matriz}")
print(f"Valor maximo del matriz: {matriz_maximo}")
print(f"Valor minimo del matriz: {matriz_minimo}")
```

```python
# maximos y minimos por filas (axis=1)
matriz_max_fila = matriz.max(axis=1)
matriz_min_fila = matriz.min(axis=1)
print(f"matriz:\n {matriz}")
print(f"Valor maximo de cada fila: {matriz_max_fila}")
print(f"Valor minimo de cada fila: {matriz_min_fila}")
```

```python
# maximos y minimos por columnas (axis=0)
matriz_max_col = matriz.max(axis=0)
matriz_min_col = matriz.min(axis=0)
print(f"matriz:\n {matriz}")
print(f"Valor maximo de cada columna: {matriz_max_col}")
print(f"Valor minimo de cada columna: {matriz_min_col}")
```

```python
m_pos_max = matriz.argmax() # indice del maximo valor
m_pos_min = matriz.argmin() # indice del minimo valor
m_pos_max_fila = matriz.argmax(axis=1) # indice del maximo valor por fila
m_pos_max_col = matriz.argmax(axis=0) # indice del maximo valor por columna

print(f"matriz:\n {matriz}")
print(f"Valor maximo global de la matriz: {matriz_maximo}, ubicado en la posicion {m_pos_max}")
print(f"Valor minimo global de la matriz: {matriz_minimo}, ubicado en la posicion {m_pos_min}")
print(f"Posicion de los maximos en cada fila: {m_pos_max_fila}")
print(f"Posicion de los minimos en cada columna: {m_pos_max_col}")
```

### METODOS `mean()`, `std()`, `sum()`

Permiten calcular el promedio, la desviacion estandar, y la suma de los elementos del arreglo.

```python
print(vector)
print(matriz)
```

```python
# suma de los elementos del vector y la matriz
print(f"Suma de todos los elementos del vector: {vector.sum()}")
print(f"Suma de todos los elementos de la matriz: {matriz.sum()}")
```

```python
print(f"Promedio por columna de la matriz: {matriz.mean(axis=0)}")
print(f"Promedio por fila de la matriz: {matriz.mean(axis=1)}")
```

```python
print(f"Desviacion estandar de la matriz: {matriz.std()}")
print(f"Desviacion estandar por columna de la matriz: {matriz.std(axis=0)}")
print(f"Desviacion estandar por fila de la matriz: {matriz.std(axis=1)}")
```
