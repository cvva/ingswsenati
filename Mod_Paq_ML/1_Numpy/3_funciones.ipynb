# Funciones `zeros()`, `ones()`, `full()`

permiten crear arreglos de cualquier tamaño con todos los valores iguales a cero (zeros), uno (ones), especifico (full)

```python
import numpy as np
```

```python
# arreglo de 3D con ceros
a = np.zeros((4, 2, 3))
print(a)
```

```python
print(f"Dimensiones del arreglo: {a.ndim}")
print(f"Forma del arreglo: {a.shape}")
print(f"Tamaño del arreglo: {a.size}")
```

```python
b = np.ones(7, dtype=np.int8)
print(b)
```

```python
b.dtype
```

```python
c = np.full((3, 3), 7)
print(c)
```

# CREACION DE ARREGLOS CON `arange()` + EL METODO `reshape()`

La funcion arange permite generar un arreglo de 1D con valores enteros en un rango especifico:

```python
# np.arange(0,10,1)
arreglo_1 = np.arange(2,20,2)
print(arreglo_1)
```

```python
arreglo_2 = np.arange(2, 15, dtype=np.int8)
print(arreglo_2)
```

```python
help(np.arange)
```

```python
arreglo_3 = np.arange(3, 31, 3)
print(arreglo_3)
```

Y podemos usar la funcion reshape para convertir el vector a un arreglo de 2 o mas dimensiones

```python
arreglo_4 = np.arange(2, 101, 2)
print(arreglo_4)
print("\nNro elementos", arreglo_4.size)
```

```python
arreglo_5 = arreglo_4.reshape(5, 10)
print(arreglo_5)
```

```python
arreglo_4.reshape(25, 2)
```

# CREACION DE ARREGLOS CON `linspace()` + EL METODO `reshape()`

La funcion `linspace()` permitir generar un arreglo de 1D con valores decimales en un rango especifico

```python
arreglo_6 = np.linspace(0, 10, 100)
print(arreglo_6)
```

```python
arreglo_7 = np.linspace(0, 10, 20)
print(arreglo_7)
```

```python
arreglo_8 = arreglo_7.reshape(4, 5)
print(arreglo_8)
```

```python
arr = np.linspace(0, 10, 100).reshape(20, 5)
print(arr)
```
