# Creacion de arreglos en Numpy

```python
#Importar la libreria Numpy
import numpy as np
```

la forma de crear arreglo usando funcion array que toma como argumento de entrada una lista de Python

```python
#creacion de arreglo 1D
a = np.array([1,2,3,4])
b = np.array([6.2,7.3,8.1,9.4])
print(a)
print(b)
```

```python
#ambos arreglos son de tipo "ndarray"
print(type(a))
print(type(b))
```

```python
d = np.array([7,9.2,'hola',True])
print(d)
print(type(d))
```

```python
# Arreglo de 2D (matriz)
arr = np.array([[1,2,3,7],[4,5,6,8]])
print(arr)
```

```python
print(type(arr))
```
