## OPERADORES DE ARREGLOS

Los operadores en Numpy permiten realizar operaciones secillas sobre cada uno de los elementos de un arreglo

### OPERADORES ARITMETICAS CON ESCALARES

Podemos operar los operadores simples en Python (con `+`, `-`, `*`, `**`) para realizar operaciones aritmeticas entre un arreglo y un escalar

```python
import numpy as np

arreglo = np.arange(5, 15)
print(arreglo)
```

```python
arreglo + 2
```

```python
arreglo - 2
```

```python
print(arreglo)
arreglo += 2
print(arreglo)
```

```python
arreglo * 0.5
```

```python
arreglo ** 2
```

### OPERACIONES ARITMETICAS ENTRE ARREGLOS

Para realizar operacones de este tipo ambos arreglos deben tener el mismo tamaño:

```python
arreglo_1 = np.linspace(3,5,10)
arreglo_2 = np.linspace(7,9,10)
arreglo_3 = np.linspace(10,15,12)
print(arreglo_1)
print(arreglo_2)
print(arreglo_3)
```

```python
arreglo_1 + arreglo_2
```

```python
arreglo_1 * arreglo_2
```

```python
arreglo_1 * arreglo_3
```

```python
# 
matriz_1 = arreglo_1.reshape(5,2)
matriz_2 = arreglo_2.reshape(5,2)
print(matriz_1)
print(matriz_2)
```

```python
matriz_multip = matriz_1 * matriz_2
print(f"Tamaño de la matriz resultante: {matriz_multip.size}")
print(f"Estructura de la matriz resultante: {matriz_multip.shape}")
print(matriz_multip)
```

### OPERACONES LOGICAS

Se puede aplicar operadores de comparacion para comparar cada elemento del arreglo con un escalar o elemento a elemento (en el caso de comparar dos arreglos)

```python
arr_1 = np.linspace(5,50,20)
print(arr_1)
```

```python
print(arr_1 >= 25)
```

```python
arr_2 = np.linspace(4,60,20)
print(arr_2)
```

```python
arr_2 > arr_1
```

```python
arr_2 != arr_1
```
